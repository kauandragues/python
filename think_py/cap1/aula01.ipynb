## Inteiros x Flutuantes
Em Python existem dois tipos de números: int e float.
- int: representa um número inteiro.
- float: representa um número com ponto decimal.

Somar, subtrair e multiplicar números inteiros sempre resultará em um número inteiro. Porém, se dividir dois números inteiros, o resultado será um número flutuante.

```python
83 // 2 
# Esse é um operador de divisão inteira, ele sempre apresentará o resultado inteiro da divisão.
# Ele também é conhecido como floor divison, pois sempre arredonda o resultado da divisão para baixo.
```

```python
-5 // 2 # -5 / 2 = 2.5 -> -2.5 arredondado para baixo é -3
```

```python
7 ** 2 # operador de exponenciação, 7 elevado à 2
```

# Operadores Bit a Bit
São operadores que servem para fazer cáculos utilizando os bits de números. Podem ser úteis para resolver alguns problemas clássicos de LeetCode além de servirem para manipular dados em baixo nível.

```python
13 & 5 # AND

#   1 1 0 1  (13)
# & 0 1 0 1  (5)
# ----------
#   0 1 0 1  → Em decimal, isso vale **5**
```

```python
13 | 5 # OR

#   1 1 0 1  (13)
# | 0 1 0 1  (5)
# ----------
#   1 1 0 1  → Em decimal, isso vale **13**
```

```python
13 ^ 5 # XOR

#   1 1 0 1  (13)
# ^ 0 1 0 1  (5)
# ----------
#   1 0 0 0  → Em decimal, isso vale **8**
```

```python
~5 # NOT

# Complemento de 2
# 5 -> 0101 -> inverte os bits -> 1010 -> soma +1 -> 1011 -> primeiro bit é 1, ou seja, é negativo. E os outros bits compoem o número de verdade -> 1 011 -> -6 
```

```python
5 << 2 # Deslocamento para esquerda 

# Mesma coisa que multiplicar por 2 ** n
# 0101 -> 1010 -> 10100 -> 20
```

```python
5 >> 2 # Deslocamento para direita

# Mesma coisa que dividir por 2 ** n
# 0101 -> 0010 -> 0001 -> 1
```

# Linguagens formais e naturais
- Naturais: são os idiomas que as pessoas falam (português, inglês, espanhol...), elas surgiram naturalmente.
- Formais: são cojuntos de regras e signos que representam algo com alguma finalidade específica. Linguagens de Programação foram criadas por seres humanos para expressar operações computacionais.

Elas se diferenciam ainda mais em 3 pontos:
- Ambiguidade: mais na natural e menos na formal.
- Redundância: mais na natural e menos na formal.
- Literalidade: mais na formal e menos na natural.
